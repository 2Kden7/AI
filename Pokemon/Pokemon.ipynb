# Question:
## How does bulbasaurs hp compare to the rest of the pokemon?
# Study Type:
### This study will be a census study because we are comparing data of ALL the pokemon that have been released, and census studies compare data from an entire population. 

```python
import pandas as pd
import seaborn as sns

#Variables 
pokemon = pd.read_csv("pokemon.csv")
bulbasaur_hp = pokemon.loc[pokemon['english_name'] == 'Bulbasaur', 'hp'].values[0]
print(f"Bulbasaur's HP is {bulbasaur_hp}")
mean_hp = pokemon['hp'].mean()
std_hp = pokemon['hp'].std()

#Calculate the Z-Score
z_score = (bulbasaur_hp - mean_hp) / std_hp
print(f"Bulbasaur's HP Z-Score is: {z_score}")

#Visual graphs 
sns.histplot(data=pokemon, x="hp")
```

```python
sns.boxplot(data=pokemon, x="hp")
```

# Conclusion:
### Given the histogram, boxplot, and looking at bulbasaurs z-score, we can determine that his hp is below average compared to other pokemon that have been released. Specifically -0.9456775627585371 below the standard deviation
