# Question: 
## Does smoking cigarettes determine if you eat steak?
# Study Type:
### This would be an observational study due to the fact that we are only using data that 560 people in the United States have answered, not the entire population.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

steak = pd.read_csv("steak.csv")

#This is my coding segment where it shows you how many people smoke and how many eat steak
cigarettes = steak[steak["Do you ever smoke cigarettes?"] == "Yes"]
print("People who smoke cigarettes:", len(cigarettes))
steaktwo = steak[steak["Do you eat steak?"] == "Yes"]
print("People who eat steak:", len(steaktwo))

#This is my coding segment where I calculate if someone does both
do_both = steak[
    (steak["Do you ever smoke cigarettes?"] == "Yes") &
    (steak["Do you eat steak?"] == "Yes")
    ]
print("People who do both:", len(do_both))

sns.countplot(data=steak, x="Do you ever smoke cigarettes?")
plt.xlabel("Smoke cigarettes?")
plt.ylabel("Number of respondents")
plt.title("Cigarette Smoking Responses")
plt.show()

sns.countplot(data=steak, x="Do you eat steak?")
plt.xlabel("Eat Steak?")
plt.ylabel("Number of respondents")
plt.title("Steak eating responses")
plt.show()
```

# Data Analysis 
### Given the data and the bar graph, we can determine that there is a correlation between if you smoke and eat steak. We can see that only 84 people who were surveyed smoked, but of those 84 people, 72 of them also ate steak. This means that if you smoke then you are also more likely to eat steak as well.
