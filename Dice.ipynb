### My Question
***If we roll a die 1000 times, how many times will it land on 5?***

### My Study
#### The random process is rolling a die 1000 times to see how many times it lands on 5.
#### One trial will look like us programming the code and "rolling the die" and there are 6 possible outcomes because we are using a six-sided die.
#### The event we are interested in is which number the die lands on. 


```python
### My Analysis
# First, write the code you need to simulate rolling your die 1000 times.
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

roll_die = pd.Series(np.random.randint(1, 7, size=1000))
counts = roll_die.value_counts().sort_index()
print("Results after 1,000 rolls:")
print(counts)

# Then, write the code you need to estimate a probability from the simulated data.
prob_5 = (roll_die == 5).mean()
print(prob_5)
# Finally, write the code you need to calculate a theoretical probability.
favorable_outcomes = 1
total_outcomes = 6
theoretical_probability = favorable_outcomes / total_outcomes
print(theoretical_probability)

# Add as many cells as you need for your analysis
```

```python
empirical = roll_die.value_counts(normalize=True).sort_index()

df = pd.DataFrame({
    'Empirical': empirical,
    'Theoretical': 1/6
})

df.plot(kind='bar', figsize=(8, 5))
plt.title('Empirical vs. Theoretical Probability of Die Rolls')
plt.xlabel('Die Face')
plt.ylabel('Probability')
plt.ylim(0, 0.25)
plt.xticks(rotation=0)
plt.legend()
plt.show()
```

## My Conclusion
#### Theoretical Probability is 0.1667 and empirical probability is 0.172. This means theoretically, we should land on 5 about 0.1667 (16.67% of the time), and in reality we landed on 5 about 0.172 (17.2%) of the time. 
### This gives us the data in a visualized way. We can see how similar both the empirical and theoretical probability was for rolling a 5.
