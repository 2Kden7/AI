# Question
## How do CU Boulder's and Colorado School Of Mines ACT scores compare to all post-secondary institutions in the United States?

# Study Type
## This study type will be an observational study because some schools in the United States don't post their ACT scores, therefore we can't analyze all the data in a dataset. 

```python
import pandas as pd
import seaborn as sns
#import matplotlib.plyplot as plt

#Reading the dataset in
college = pd.read_csv("cleaned_college.csv")

#Mean, Standard Deviation, and Median for all colleges in the dataset
mean_act = college["act_midpoint"].mean()
std_act = college["act_midpoint"].std()
median_act = college["act_midpoint"].median()
print(mean_act, median_act, std_act)

#CU Boulder variables + z-score calculation
cuboulder = college[college["school_name"] == "University Of Colorado Boulder"].copy()
target_act = college.loc[
    college["school_name"] == "University of Colorado Boulder", "act_midpoint"
].values[0]
cuz_score = (target_act - mean_act) / std_act
print(cuz_score)

#Colorado School of Mines variables + z-score calculation
mines = college[college["school_name"] == "Colorado School of Mines"].copy()
target_act2 = college.loc[
    college["school_name"] == "Colorado School of Mines", "act_midpoint"
].values[0]
minesz_score = (target_act2 - mean_act) / std_act
print(minesz_score)

sns.histplot(data=college, x="act_midpoint")
```

## Analysis

### After we found the z-scores of both CU Boulder and the Colorado School of Mines, we can determine where they stand compared to other post-secondary institutions. We found that the z-score of CU Boulder is 0.8376253377010353. This means that they are 0.84 (rounded) standard deviations above the Mean ACT scores of all post-secondary institutions. After finding the z-score of the School of Mines, which was 1.5524317109297776, we can determine that they are 1.55 (rounded) standard deviations above the Mean ACT scores of all post-secondary institutions.
