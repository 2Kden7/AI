# Question:
### How many mushrooms are edible and how many are poisonous?

#### I will be using a census study to answer this question because we have access to the total population of mushrooms and we have the data to determine which ones are poisonous vs. edible. This can also be answered with frequency because we can find out how many mushrooms are edible by counting them.

# Anaylsis:

```python
import pandas as pd
import seaborn as sns
fungi = pd.read_csv("mushrooms.csv")

poisonousMushroomClass = (fungi[fungi["class"] == "p"])
count1 = len(poisonousMushroomClass)
print("This is how many mushrooms are poisonous: ", count1)

edibleMushroomClass = (fungi[fungi["class"] == "e"])
count2 = len(edibleMushroomClass)
print("This is how many mushrooms are edible: ", count2)

fungi["class"].value_counts(normalize=True)
```

```python
sns.countplot(data=fungi, x="class")
```

# Final Conclusion:
### According to the dataset provided, there are 3916 poisonous mushrooms and 4208 edible mushrooms. The frequency of poisonous (p) is 3916 and the frequency of edible (e) is 4208. The relative frequency of poisonous is around 48% (0.482029) and the relative frequency of edible is around 52% (0.517971).
