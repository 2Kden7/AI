# Question:
## What's the average amount of streams of all the songs listed in the data?
# Study type:
### This would be an observational study because we are only studying a certain amount of songs, not EVERY song on spotify (which would be extremely diffiult).

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

spotify = pd.read_csv("spotify.csv", encoding='latin-1')
spotify["streams"] = pd.to_numeric(spotify["streams"].squeeze(), errors="coerce")
median_streams = spotify["streams"].median()
print(median_streams)

sns.histplot(data=spotify, x="streams",bins=20)
plt.xlabel("Spotify streams")
plt.ylabel("Number of songs")
plt.title("Distribution of Spotify streams")
plt.show()
```

# Conclusion:
### The average amount of streams for spotify songs using the given data is 290530915.0. We can also see that the histogram is skewed to the right.
